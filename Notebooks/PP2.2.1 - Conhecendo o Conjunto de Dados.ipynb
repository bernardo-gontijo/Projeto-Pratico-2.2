#### Histogramas e Boxplots dos Atributos Preditores:

```python
atributos = df.columns.drop('Class')

sns.set_theme(style='whitegrid')
fig = plt.figure(figsize=(20, 22), layout='constrained')
subfiguras = fig.subfigures(nrows=4, ncols=4, wspace=0.03, hspace=0.04)

for atributo, subfig in zip(atributos, subfiguras.flat):
    ax_hist, ax_box = subfig.subplots(nrows=2, sharex=True, height_ratios=[3, 1])

    media = df[atributo].mean()
    mediana = df[atributo].median()

    # Histogramas
    sns.histplot(df[atributo], bins=50, color='steelblue', ax=ax_hist)
    ax_hist.axvline(media, color='red', linestyle='--', linewidth=1.5, label=f'Média: {media:.4g}')
    ax_hist.axvline(mediana, color='green', linewidth=1.5, label=f'Mediana: {mediana:.4g}')
    ax_hist.set_ylabel('Frequência')
    ax_hist.legend(fontsize=9)

    # Boxplots
    sns.boxplot(x=df[atributo], color='lightsteelblue', fliersize=2, ax=ax_box)
    ax_box.set_xlabel('')

    subfig.suptitle(atributo, fontsize=13)

fig.suptitle('Distribuição dos 16 atributos preditores (histograma + boxplot)', fontsize=18, fontweight='bold')
plt.show()
```

#### Gráfico distribuição Atributo-Alvo:

```python
contagem = df['Class'].value_counts()
percentual = df['Class'].value_counts(normalize=True) * 100

fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(x=contagem.index, y=contagem.values, color='steelblue', alpha=0.9, ax=ax)

for i, (qtd, pct) in enumerate(zip(contagem, percentual)):
    ax.text(i, qtd + 40, f'{qtd}\n({pct:.1f}%)', ha='center', va='bottom', fontsize=10, bbox=dict(facecolor='white', pad=1))

ax.set_title("Distribuição do atributo-alvo (Class)", fontsize=14)
ax.set_xlabel('Variedade')
ax.set_ylabel('Número de exemplos')
ax.set_ylim(0, contagem.max() * 1.18)
plt.tight_layout()
plt.show()
```

#### Balanceamento do Atributo-Alvo:

O atributo-alvo `Class` é desbalanceado, para que as classes fossem igualmente distribuidas seriam algo em torno de 2000 exemplos para cada classe. Como o gráfico de distribuição mostra a classe **DERMASON** é majoritária na base de dados enquanto a classe **BOMBAY** é minoritária, tendo uma razão de quase 7 : 1. Porém a distribuição das classes intermediárias está proxima de um valor ideal para o treinamento, podendo considerar um desbalanceamento moderado no geral. Por isso, futuramente será interessante estretificar as partições de treino e teste.

**3. Que tipo de tarefa de Aprendizado de Máquina é sugestiva para este problema?**

Classificação Multiclasse Supervisionada

**3.1. Métricas de desempenho compatíveis para a avaliação do problema**

Acurácia (pode ser enganosa) e Acurácia Balanceada, Precisão, Recall, F1-Score. Essas métricas sendo macro para levar em conta todas as clsses, para não serem "enganadas" pelo desbalanceamento.

**3.2. Métrica escolhida pela equipe**

F1-Score Macro será nossa principal, pois ela combina outras métricas (Recall e Precisão) e nela todas as classes têm o mesmo peso agindo contra confusões por conta do desbalanceamento.
